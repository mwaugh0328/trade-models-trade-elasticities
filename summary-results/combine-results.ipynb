### Combine Estimation Results

This notebook reads CSV files with date stamps (e.g., `-12726`) from the results folder and combines them into a single DataFrame.

```python
import pandas as pd
import glob
import re
from pathlib import Path
```

```python
# Get all CSV files in the results folder
results_folder = Path('../results')
all_csv_files = list(results_folder.glob('*.csv'))

# Filter to only include files with a date pattern (e.g., -12726)
# Pattern: filename containing a hyphen followed by digits before .csv
date_pattern = re.compile(r'-\d+(-[\w-]+)?\.csv$')
dated_files = [f for f in all_csv_files if date_pattern.search(f.name)]

# Exclude melitz sigma-theta files and model2 files
dated_files = [f for f in dated_files if 'sigma-theta' not in f.name]
dated_files = [f for f in dated_files if 'model2' not in f.name]

# Only include files with the date 12726
dated_files = [f for f in dated_files if '12726' in f.name]

print(f"Found {len(dated_files)} files with date stamps:")
for f in dated_files:
    print(f"  - {f.name}")
```

```python
# Separate files into exact and over categories
exact_files = [f for f in dated_files if 'exact' in f.name]
over_files = [f for f in dated_files if 'over' in f.name]

print(f"Exact files ({len(exact_files)}):")
for f in exact_files:
    print(f"  - {f.name}")

print(f"\nOver files ({len(over_files)}):")
for f in over_files:
    print(f"  - {f.name}")
```

```python
# Read and combine EXACT files
exact_dfs = []
for file in exact_files:
    df = pd.read_csv(file)
    df['source_file'] = file.name
    exact_dfs.append(df)

exact_df = pd.concat(exact_dfs, ignore_index=True)

# Read and combine OVER files
over_dfs = []
for file in over_files:
    df = pd.read_csv(file)
    df['source_file'] = file.name
    over_dfs.append(df)

over_df = pd.concat(over_dfs, ignore_index=True)

# Set display options to show 2 decimal places
pd.set_option('display.float_format', '{:.2f}'.format)

print(f"Exact DataFrame: {len(exact_df)} rows")
print(f"Over DataFrame: {len(over_df)} rows")
```

```python
# Display EXACT results
print("=== EXACT Estimation Results ===")
exact_df.sort_values(by = ["θ"], inplace=True)

exact_df
```

```python
# Save the combined results to CSV
exact_df.to_csv('combined-exact-results.csv', index=False, float_format='%.2f')
over_df.to_csv('combined-over-results.csv', index=False, float_format='%.2f')
print("Saved: combined-exact-results.csv")
print("Saved: combined-over-results.csv")
```

```python
# Display OVER results
print("=== OVER Estimation Results ===")
over_df
```

```python
# Combine BEJK sigma files
# Pattern: bejk-estimate-{exact|over}-{date}-sigma{value}.csv
sigma_pattern = re.compile(r'sigma(\d+)\.csv$')

bejk_sigma_files = [f for f in results_folder.glob('bejk-estimate-*-sigma*.csv')]

bejk_sigma_dfs = []
for file in bejk_sigma_files:
    df = pd.read_csv(file)
    df['source_file'] = file.name
    # Extract sigma value from filename (e.g. sigma15 -> 1.5)
    m = sigma_pattern.search(file.name)
    df['sigma'] = int(m.group(1)) / 10 if m else None
    # Extract identification type (exact or over)
    df['identification'] = 'exact' if 'exact' in file.name else 'over'
    bejk_sigma_dfs.append(df)

bejk_sigma_df = pd.concat(bejk_sigma_dfs, ignore_index=True)

# Split into exact and over
bejk_sigma_exact_df = bejk_sigma_df[bejk_sigma_df['identification'] == 'exact'].sort_values(by='sigma').reset_index(drop=True)
bejk_sigma_over_df  = bejk_sigma_df[bejk_sigma_df['identification'] == 'over'].sort_values(by='sigma').reset_index(drop=True)

bejk_sigma_exact_df.to_csv('combined-bejk-sigma-exact-results.csv', index=False, float_format='%.2f')
bejk_sigma_over_df.to_csv('combined-bejk-sigma-over-results.csv', index=False, float_format='%.2f')
print(f"Saved combined-bejk-sigma-exact-results.csv ({len(bejk_sigma_exact_df)} rows)")
print(f"Saved combined-bejk-sigma-over-results.csv ({len(bejk_sigma_over_df)} rows)")

print("\n=== BEJK Sigma — Exact ===")
display(bejk_sigma_exact_df)
print("\n=== BEJK Sigma — Over ===")
display(bejk_sigma_over_df)
```

```python
# Combine Krugman model2 files
krugman2_exact_df = pd.read_csv(results_folder / 'krugman-model2-estimate-exact-12726.csv')
krugman2_over_df  = pd.read_csv(results_folder / 'krugman-model2-estimate-over-12726.csv')

krugman2_exact_df.to_csv('combined-krugman-model2-exact-results.csv', index=False, float_format='%.2f')
krugman2_over_df.to_csv('combined-krugman-model2-over-results.csv', index=False, float_format='%.2f')
print(f"Saved combined-krugman-model2-exact-results.csv ({len(krugman2_exact_df)} rows)")
print(f"Saved combined-krugman-model2-over-results.csv ({len(krugman2_over_df)} rows)")

print("\n=== Krugman Model2 — Exact ===")
display(krugman2_exact_df)
print("\n=== Krugman Model2 — Over ===")
display(krugman2_over_df)
```
